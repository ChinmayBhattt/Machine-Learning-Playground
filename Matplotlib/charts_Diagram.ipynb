### Data Visualization Tools and Their Purpose
- Bar Charts - Catagory Comparison
- Pie Charts - Proportion Illustration
- Histograms - Numerical Distribution

```python
import matplotlib.pyplot as plt
```

#### Bar Charts

```python
# plt.bar(x, height, color, width, label)

product = ['A','B','C','D']
sales = [1000, 1500, 800, 1200]

t = plt.bar(product,sales, color = 'green', label = 'Sales 2025')
plt.xlabel('Product')
plt.ylabel('Sales')
plt.title('2025 Product Sales Comparison')
plt.legend()
```

```python
#horizontal bars and vertical bars - barh()
plt.barh(product,sales, color = 'red', label = 'Sales 2025')
plt.legend(loc = 'lower right')
```

#### Pie Charts

```python
# plt.pie(values, colors, labels, autopct=%1.1f%%)

regions = ['North','South','East','West']
revenue = [3000, 2000, 1500, 1000]
# autopercentage
plt.pie(revenue,labels=regions, autopct='%1.1f%%', colors=['gold','skyblue','lightgreen', 'coral']) 
plt.title('Revenue Contribution by Region')
plt.show()
```

#### Histograms

```python
# plt.hist(data, bins=number_of_bins, color, edgecolor(border color))
scores = [45,57,24,39,33,11,23,56,86,90,90,80,58,94,90,90]
plt.hist(scores, bins = 5, color = 'purple', edgecolor = 'black')
plt.xlabel('Score Range')
plt.ylabel('Number of Student')
plt.title('Score Distibution of Students')
```
