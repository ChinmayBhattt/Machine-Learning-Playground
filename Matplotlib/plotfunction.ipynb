```python
import matplotlib.pyplot as plt
```

```python
months = [1,2,3,4]
sales = [100,1500,1200,1800]

plt.plot(months,sales, color='red', linestyle='--', linewidth = 2, marker = 'o', label = '2025 Sales Data')
plt.xlabel('Months')
plt.ylabel('Sales Per Month')
plt.title('Monthly Sales Data Report')
# plt.legend()
plt.legend(loc = 'upper left') #upper left -> top left corner 
plt.grid(color = 'darkgreen', linestyle = ':', linewidth = 1)
plt.xlim(1, 4)
plt.ylim(0, 2000)
plt.xticks([1,2,3,4], ['M1','M2','M3','M4'])
plt.show()
plt.close()
```
