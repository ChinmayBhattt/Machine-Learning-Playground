```python
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4]
y = [10,20,15,25]

plt.subplot(1,2,1)
plt.title("Line Chart")
plt.plot(x,y)

plt.subplot(1,2,2)
plt.title("Bar Chart")
plt.bar(x,y)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(1,2, figsize=(10,5))

x = [1,2,3,4]
y = [10,20,15,25]

ax[0].plot(x,y)
ax[0].set_title("Line Chart")

ax[1].bar(x,y)
ax[1].set_title('Bar Chart')

plt.show()
```
