```python
import matplotlib.pyplot as plt

x = [1,2,3,4]
y = [10,15,20,25]

#create plot

plt.plot(x,y, color = 'blue', marker = 'o')
plt.title('Simple LIne Plot')
plt.xlabel('X axis')
plt.ylabel('Y axis')

plt.savefig('line_plot.png', dpi = 300, bbox_inches = 'tight')
plt.show()
```
