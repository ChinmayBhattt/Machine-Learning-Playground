```python
import matplotlib.pyplot as plt
```

```python
# plt.scatter(x,y,color,marker, label)

hours_study = [1,2,3,4,5,6,7,8]
exam_scores = [50 ,55, 60, 65, 70, 75, 80, 85]

plt.scatter(hours_study, exam_scores, color = 'green', marker = '^', label = 'Student Data')
plt.xlabel('Hours Study')
plt.ylabel('Exam SCore')
plt.title('Relationship Between Study Tie and Exams Score')
plt.legend()
plt.grid()
```

```python
plt.scatter([1,2,3], [50,60,70], color = 'blue', label = 'Class A') # Group 1
plt.scatter([1,2,3], [45,50,52], color = 'orange', label = 'Class B') # Group 2


plt.xlabel('Hours Study')
plt.ylabel('Exam SCore')
plt.title('Comparison of Two Classes')
plt.legend()
plt.grid()
```
