```python
import numpy as np

marks = np.array([
    [65, 70, 75],   # Student 1
    [55, 60, 58],   # Student 2
    [90, 92, 88],   # Student 3
    [30, 25, 40],   # Student 4
    [50, 45, 55]    # Student 5
])
```

```python
# student_marks = int(input("Enter a Student's Marks: "))

for i in range(5):
    print(f"Student {i+1}")
    print(f"Maximing Marks:",marks[i].max())
    print(f"Minimum Marks:",marks[i].min())
    print(f"Average Marks:",marks[i].mean())
    print("\n")
    
```
